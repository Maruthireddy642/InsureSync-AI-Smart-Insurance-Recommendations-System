```python
#Univariate Exploration

plt.figure(figsize=(10,6))

# First subplot - Age Distribution
plt.subplot(3,2,1)
sns.histplot(data=data, x='Age', bins=16, color='lightblue', edgecolor='black')
plt.title('Age wise Distribution of Patients', fontweight='bold')
plt.xlabel('Age', size=13)
plt.ylabel('Count of Patients', size=13)

# Second subplot - Gender Distribution
plt.subplot(3,2,2)
color = ['lightblue','skyblue']
sns.countplot(data=data, x='Gender',palette=color, edgecolor='black')
plt.title('Gender wise Distribution', fontweight='bold')
plt.xlabel('Gender', size=13)
plt.ylabel('Count of Patients', size=13)

# Third subplot - Medical condition Distribution
plt.subplot(3,2,3)
sns.countplot(data=data,x='Medical Condition',edgecolor='black',color='lightblue')
plt.title('Medical Condition wise Distribution',fontweight='bold')
plt.xlabel('Medical Condition',size=13)
plt.ylabel('Count of Patients',size=13)

# Fourth subplot - Insurance Provider Distribution
plt.subplot(3,2,4)
sns.countplot(data=data,x='Insurance Provider',edgecolor='black',color='lightblue')
plt.title('Insurance Provider wise Distribution',fontweight='bold')
plt.xlabel('Insurence Provider',size=13)
plt.ylabel('Count of Patients',size=13)

# Fifth subplot - Blood Type Distribution
plt.subplot(3,2,5)
sns.countplot(data=data,x='Blood Type',edgecolor='black',color='lightblue')
plt.title('Blood Type wise Distribution',fontweight='bold')
plt.xlabel('Blood Type',size=13)
plt.ylabel('Count of Patients',size=13)

# Sixth subplot - Test Results Distribution
plt.subplot(3,2,6)
sns.countplot(data=data,x='Test Results',edgecolor='black',color='lightblue')
plt.title('Test Results wise Distribution',fontweight='bold')
plt.xlabel('Test Results',size=13)
plt.ylabel('Count of Patients',size=13)

plt.tight_layout()
plt.show()
```

```python
#Shows gender distribution per medical category

plt.figure(figsize=(10,5))
colors=['#4F81BD','#C0504D']
sns.countplot(data=data,x='Medical Condition',hue='Gender',palette=colors,edgecolor='black')
plt.title('Gender Distribution by Medical Category',fontweight='bold')
plt.xlabel('Medical Condition',size=13)
plt.ylabel('Count of Patients by Gender',size=13)
plt.show()
```

```python
#Shows age distribution per medical category

plt.figure(figsize=(10,5))
sns.countplot(data=data,x='Medical Condition',hue='Age')
plt.legend([],[], frameon=False)  
plt.title('Age Distribution by Medical Category',fontweight='bold')
plt.xlabel('Medical Condition',size=13)
plt.ylabel('Count of Patients by Age group',size=13)
plt.show()
```

```python
#Shows co-relation between Blood type & Age

plt.figure(figsize=(10,5))
sns.violinplot(data=data, x='Blood Type', y='Age',color='skyblue',linecolor='black')
plt.title('Blood Type vs Age Distribution', fontweight='bold')
plt.xlabel('Blood Type', size=13)
plt.ylabel('Age', size=13)
plt.show()
```

```python
#Shows age wise admission type

plt.figure(figsize=(10,5))
sns.lineplot(data=data,x='Admission Type',y='Age')
plt.title('Age-wise Admission Type',fontweight='bold')
plt.xlabel('Admission Type',size=13)
plt.ylabel('Age Distribution',size=13)
plt.show()
```

📊 Multivariate Analysis

```python
#Shows Age Distribution by Medical Category & Gender

plt.figure(figsize=(10,5))
sns.barplot(data=data,x='Medical Condition',y='Age',hue='Gender',edgecolor='black',palette='Set2')
plt.title('Age distribution by Medical Condition & Gender',fontweight='bold')
plt.xlabel('Medical Condition',size='13')
plt.ylabel('Age',size=13)
plt.show()
```

```python
#Shows Co-relation of Average Age with Blood type & Medical condition

pivot_table = data.pivot_table(
    values='Age', 
    index='Blood Type', 
    columns='Medical Condition', 
    aggfunc='mean'  
)

plt.figure(figsize=(9,3))
sns.heatmap(pivot_table, annot=True, cmap="magma", linewidths=0.5)
plt.title("Average Age by Blood Type & Medical Condition", fontweight="bold")
plt.xlabel('Medical Condition',size=13)
plt.ylabel('Blood Type',size=13)
plt.show()
```

### **1 Univariate Analysis**
* **Age** → Patients are fairly evenly distributed across age groups 20–80.
* **Gender** → Male and female patients are almost equally represented.
* **Medical Condition** → All listed conditions have near-equal patient counts.
* **Insurance Provider** → Each major provider covers a similar number of patients.
* **Blood Type** → Blood groups are evenly distributed across patients.
* **Test Results** → Normal, Inconclusive, and Abnormal results are balanced.

### **2 Bivariate Analysis**
* Gender is evenly distributed throughtout Medical conditions with male slightly higher.
* Age is evenly distributed throghout diffrent Mediacal Conditions.
* Age is evenly distributed throughout diffrent Blood Types

### **3 Multivariate Analysis**
* Age is evenly distributed throughout Gender & Medical conditions but males are slightly higher in Cancer & Arthritis and women in Hypertension.
* Across all Blood types & Medical Conditions the average age consistently fa;;s under a narrow range(50-53).
