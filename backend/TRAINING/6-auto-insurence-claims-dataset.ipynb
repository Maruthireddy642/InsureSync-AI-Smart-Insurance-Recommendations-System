## Colummn Summary
* **months_as_customer:** Number of months the customer has been with the insurance company.
* **age:** Age of the insured.
* **policy_number:** Unique identifier for each insurance policy.
* **policy_bind_date**: Date when the insurance policy was initiated or bound.
* **policy_state:** State where the insurance policy is issued.
* **policy_csl**: Coverage limit of the insurance policy.
* **policy_deductable:** Deductible amount for the insurance policy.
* **policy_annual_premium:** Annual premium amount for the insurance policy.
* **umbrella_limit**: Coverage limit provided by an umbrella insurance policy.
* **insured_zip:** ZIP code of the insured's residence.
* **insured_sex** Gender of the insured.
* **insured_education_level:** Education level of the insured.
* **insured_occupation:** Occupation of the insured.
* **insured_hobbies:** Hobbies or interests of the insured.
* **insured_relationship:** Relationship status of the insured.
* **capital-gains:** Amount of capital gains.
* **capital-loss:** Amount of capital loss.
* **incident_date:** Date when the incident (e.g., accident) occurred.
* **incident_type:** Type of incident (e.g., collision, theft).
* **collision_type**: Type of collision (e.g., rear-end, side-swipe).
* **incident_severity:** Severity of the incident (e.g., minor, major).
* **incident_state**: State where the incident occurred.
* **incident_city:** City where the incident occurred.
* **incident_location:** Specific location where the incident occurred.
* **incident_hour_of_the_day**: Hour of the day when the incident occurred.
* **number_of_vehicles_involved:** Number of vehicles involved in the incident.
* **property_damage:** Whether property damage occurred in the incident.
* **bodily_injuries:** Number of bodily injuries in the incident.
* **witnesses:** Number of witnesses to the incident.
* **police_report_available: **Whether a police report is available for the incident.
* **total_claim_amount:** Total claim amount for the incident.
* **injury_claim:**m: Claim amount for property damage.
* **vehicle_claim**: Claim amount for vehicle damage.
* **auto_make:** Make of the insured vehicle.
* **auto_model**: Model of the insured vehicle.
* **auto_year:** Year of the insured vehicle.
* **fraud_reported:** Whether fraud was reported for the incident.

#  Scatter plot of Age vs Auto_make

# Q.Relation between Age vs Annual Premium 

```python
# Calculate the mean of the policy annual premium
mean_premium = data['policy_annual_premium'].mean()
sns.scatterplot(data=data, x='age', y='policy_annual_premium')
plt.axhline(mean_premium, color='red', linestyle='--', label=f'Mean Premium: {mean_premium:.2f}')

plt.title('Age vs. Policy Annual Premium')
plt.xlabel('Age')
plt.ylabel('Policy Annual Premium')
plt.show()
```

**This type of plot helps in identifying potential correlations or trends between the two variables.**

scatter plot of 'age' vs. 'policy_annual_premium' with a red dashed line indicating the average policy annual premium.

# Q. Claim Amount Distribution by Incident Type


```python
plt.figure(figsize=(10, 5))
sns.boxplot(data=data, x='incident_type', y='total_claim_amount')
plt.title('Claim Amount Distribution by Incident Type')
plt.xlabel('Incident Type')
plt.ylabel('Total Claim Amount')
plt.show()
```

# Q. Showing the Distribution of claim type based on their total amount

```python

# Calculate the total claim amounts for each type
injury_total = data['injury_claim'].sum()
property_total = data['property_claim'].sum()
vehicle_total = data['vehicle_claim'].sum()

# Data for the pie chart
claim_types = ['Injury Claim', 'Property Claim', 'Vehicle Claim']
claim_totals = [injury_total, property_total, vehicle_total]

# Create a pie chart
plt.figure(figsize=(8, 6))
plt.pie(claim_totals, labels=claim_types, autopct='%1.1f%%', startangle=140)
plt.title('Distribution of Claim Types')
plt.axis('equal')  # Equal aspect ratio ensures that pie is drawn as a circle
plt.show()
```

# Q. Police report vechicle 
**Data shows how many vechicles claims done or police report available 

```python
police_vehicle_claim = data.groupby('police_report_available')['vehicle_claim'].mean().reset_index()

# Plot the grouped bar plot
plt.figure(figsize=(8, 6))
sns.barplot(data =police_vehicle_claim, x='police_report_available', y='vehicle_claim')
plt.title('Vehicle Claim vs Police Report Availability')
plt.xlabel('Police Report Available')
plt.ylabel('Average Vehicle Claim Amount')
plt.tight_layout()
plt.show()
```
